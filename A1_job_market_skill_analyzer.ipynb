2) Add New Job Postings (interactive)

Ask the user how many job postings they want to add. For each posting, collect the role, company, and skills (skills provided as a single string, space-separated or comma-separated).
 Requirements:
  ● Job_ID should auto-increment from the maximum existing ID.
  ● Convert numeric inputs (if any) to integers where applicable.
  ● Append new values to the lists in job_data.
  
  Example interaction

How many job postings do you want to add? 2

Enter role: Junior Data Analyst

Enter company: BrightData

Enter skills (space-separated or comma-separated): Excel, SQL, Python

```python
job_data = { "Job_ID": [1, 2, 3, 4, 5],
             "Role": ["Data Analyst", "Data Scientist", "Business Analyst", "Data Engineer", "ML Engineer"],
             "Company": ["ABC Corp", "XyZ Ltd", "DataWorks", "InfraTech", "AI Labs"],
             "Skills": [ "Excel SQL Python", "Python ML Statistics", "Excel SQL PowerBI", "Python SQL Spark", "Python ML DeepLearning"]
             }

#input data(convert integer)

n=int(input("How many job postings do you want to add?:  "))


#Job_ID should auto-increment from the maximum existing ID.

max_ID=max(job_data["Job_ID"])
new_ID=max_ID + 1
max_ID=new_ID

#Append new values to the lists in job_data.

for i in range(n):
  role=input("Enter role: ")
  company=input("Enter company: ")
  skill=input("Enter skills (space-separated or comma-separated): ")

#append datas

  job_data["Job_ID"].append(new_ID)
  job_data["Role"].append(role)
  job_data["Company"].append(company)
  job_data["Skills"].append(skill)

#update datas

print("\nUpdated job data: ")
print(job_data)
```

3) Clean & Normalize Skills (text processing)

Make a function called clean_skills() that accepts skills in string format.

   Converts text to lowercase

   Replaces commas with spaces

   Removes extra spaces (normalize whitespace)
  
   Optionally removes punctuation (you may use basic str.replace() or re module)

  Apply clean_skills() to every entry in job_data['Skills'] and overwrite the list with cleaned
versions.

```python
#cleaning
def clean_skills(skills):
  skills=skills.lower()
  skills=skills.replace(","," ")
  skills=" ".join(skills.split())
  return skills

#calling and updating
job_data["Skills"]=[clean_skills(skill) for skill in job_data["Skills"]]

#printing
print(job_data["Skills"])
```

 4) Skill Extraction & Frequency

    From the cleaned skills list, split each skill string into individual skill tokens.

    Build a consolidated list of all skills that contains every skill token across all postings.

    Use a dictionary or other structure to
   compute skill_count={skill:frequency}.

    Print the top 5 most frequent skills(or all if<5)

   Expected print example:

     Top skills by frequency:

     Python: 5

     sql: 4

     Excel: 2

     ML: 2

     Power BI: 1

```python
#Create an empty list updated skill-i store seiya
updated_skills=[]

#thaniya skills-a pirichu adha updated_skills-la append pandrom
for each_skill in job_data["Skills"]:
  extracts=each_skill.split()

  for skill in extracts:
    updated_skills.append(skill)

#skill & frequency-a store panna empty {}-dictionary create pandrom
skill_count={}

#skill_count-la update pandrom......  {key:value}
for skill in updated_skills:

  if skill in skill_count:
    skill_count[skill]=skill_count[skill]+1

  else:
    skill_count[skill]=1

#skill:frequency-a base panni sort pandrom
sorted_skills=sorted(skill_count.items(),
                     key=lambda x: x[1],
                     reverse=True)

#print the skills order by frequency of condition apply
print("Top skills by frequency:\n")

for x,y in sorted_skills[:5]:
  print(f"{x}:{y}\n")
```

5) Basic Statistics & Insights

Calculate and print:

● Total number of job postings

● Number of unique skills

● Jobs that mention "python" (case-insensitive)

Sample Output

Total Jobs: 7

Unique skills: 9

Jobs mentioning Python: ['Data Analyst', 'Data Scientist', 'Data Engineer', ...]

```python
#Total number of job postings
total_jobs=len(job_data["Skills"])
print(f"Total Jobs: ",total_jobs,"\n")

#Number of unique skills, here avoid duplicates, so we can choose set().
unique_skills=set()

for skill_string in job_data["Skills"]:
  skills=skill_string.split()

  for skill in skills:
    unique_skills.add(skill)

#print(unique_skills)   #ingu unique jobs only print not print duplicate
print(f"Unique skills:" ,len(unique_skills),"\n")

#Jobs that mention "python" (case-insensitive), here accept all type of Python in only "python".
Python_jobs=[]

for i in range(len(job_data["Skills"])):
  skills=job_data["Skills"][i].lower()

  if "python" in skills:
    Python_jobs.append(job_data["Role"][i])

print(f"Jobs mentioning Python: ",Python_jobs)

```

6) Skill Search Function (functions & conditional logic)

Create a function, find_jobs_by_skill(skill_query), that returns a list of (Job_ID, Role, Company) for postings containing the skill_query token.

 ● The function must normalize the skill_query (lowercase, strip spaces)

● Use if/elif/else where appropriate

Example: find_jobs_by_skill('sql') → returns matching jobs.

```python
# create a function
def find_jobs_by_skill(skill_query):       #SQL

#The function must normalize the skill_query (lowercase, strip spaces)
  skill_query=skill_query.lower().strip()  #Normalize pandrom- sql

#Use if/elif/else where appropriate..............Example: find_jobs_by_skill('sql') → returns matching jobs.
  matching_jobs=[]

  for i in range(len(job_data["Skills"])):

    skills=job_data["Skills"][i].lower().split()    # each skill change to token. Ex-"sql"

  # matching activities
    if skill_query in skills:
      matching_jobs.append((job_data["Job_ID"][i],
                          job_data["Role"][i],
                          job_data["Company"][i]))

    elif skill_query=="":
      print(f"Please enter a skill.")
      return []

    else:
      continue

  return matching_jobs

#print
result=find_jobs_by_skill("sql")
print(result)
```

7) Lambda Filtering & List Comprehension

Use a lambda function and filter() (or a list comprehension with a lambda-like expression) to
extract job roles where the skill count (from skill_count) for any skill >= 2 — that is, roles
that include at least one relatively common skill.

Example Output

Roles with at least one common skill: ['Data Analyst', 'Data Scientist', 'ML Engineer']

```python
roles_with_common_skill=[]

for i in range(len(job_data["Role"])):

  skills=job_data["Skills"][i].lower().split()

  common_skill=filter(
      lambda skill: skill_count.get(skill,0)>=2,     #filter(condition-enna seiya,data-edhilirundhu seiya)
      skills
  )

  if list(common_skill):

    roles_with_common_skill.append(job_data["Role"][i])

print(f"Roles with at least one common skill: ",roles_with_common_skill)
```
